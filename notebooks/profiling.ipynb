### Testing the business rules

```python
expected = df["quantity"] * df["unit_price"] * (1 - df["discount_rate"])

print("Revenue matches formula:", ((df["order_revenue"] - expected).abs() < 0.01).mean())
print("Revenue includes delivery fee:", ((df["order_revenue"] - (expected + df["delivery_fee"])).abs() < 0.01).mean())
print("Max regions per city:", pd.crosstab(df["city"], df["region"]).gt(0).sum(axis=1).max())
print("Max product names per product_id:", df.groupby("product_id")["product_name"].nunique().max())
print("Max categories per product name:", df.groupby("product_name")["product_category"].nunique().max())
print("Date range:", df["order_date"].min(), "to", df["order_date"].max())
```

## CLEANING

### Fix the text columns

For each column, this removes spaces, lowercases the value, swaps it for the correct spelling from the dictionary, and fills blanks with "Unknown".

```python
clean = df.copy()

maps = {
    "gender": {"male": "Male", "m": "Male", "female": "Female", "f": "Female"},
    "customer_segment": {"regular": "Regular", "premium": "Premium", "occasional": "Occasional", "business": "Business"},
    "payment_method": {"card": "Card", "bank transfer": "Bank Transfer", "cash on delivery": "Cash on Delivery", "ussd": "USSD", "wallet": "Wallet"},
    "sales_channel": {"website": "Website", "mobile app": "Mobile App", "store": "Store", "marketplace": "Marketplace", "instagram": "Instagram"},
    "order_status": {"completed": "Completed", "cancelled": "Cancelled", "pending": "Pending", "returned": "Returned"},
    "delivery_status": {"delivered": "Delivered", "in transit": "In Transit", "delayed": "Delayed", "returned": "Returned"},
}

for col, mapping in maps.items():
    clean[col] = clean[col].str.strip().str.lower().map(mapping).fillna("Unknown")

clean["age_group"] = clean["age_group"].fillna("Unknown")
clean["customer_name"] = clean["customer_name"].str.strip()
```

```python
### Fix the data types
```

```python
clean["return_flag"] = clean["return_flag"].str.strip().str.lower().map({"yes": True, "no": False}).astype("boolean")
clean["order_date"] = pd.to_datetime(clean["order_date"])
clean["delivery_days"] = clean["delivery_days"].astype("Int64")
clean["customer_rating"] = clean["customer_rating"].astype("Int64")
clean = clean.drop(columns=["phone_number"])
```
